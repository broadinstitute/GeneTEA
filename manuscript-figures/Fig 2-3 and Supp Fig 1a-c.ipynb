# Example analysis

```python
exposed_skin_name, protected_skin_name = "Sun-Exposed Skin", "Sun-Protected Skin"
exposed_skin = pd.read_csv(FILEPATHS["sun-exposed_gtex"], sep="\t", skiprows=2)
protected_skin = pd.read_csv(FILEPATHS["sun-protected_gtex"], sep="\t", skiprows=2)
exposed_skin.shape, protected_skin.shape
```

```python
rpm = lambda x: np.log2(((x / x.sum()) * 1e6) + 1)

def find_degs(name_a, counts_a, name_b, counts_b, fdr_thresh=0.05, diff_thresh=1):
    #RPM normalized the inputs
    rpm_a, rpm_b = rpm(counts_a), rpm(counts_b)
    #find genes expressed in either sample
    expressed = (rpm_a.mean(axis=1) > 1) | (rpm_b.mean(axis=1) > 1)
    #align the two dataframes and make into lists of sample values per expressed protein-coding gene
    aligned = pd.concat({
        name_a:rpm_a.apply(lambda x: list(x.dropna()), axis=1), 
        name_b:rpm_b.apply(lambda x: list(x.dropna()), axis=1)
    }, axis=1, join="inner").loc[lambda x: x.index.isin(gsm.get_protein_coding()) & expressed.reindex(x.index)]
    #run right-tailed wilcoxon rank sums and FDR correct
    ranksum_res = aligned.apply(
        lambda x: pd.Series(scipy.stats.ranksums(x[name_a], x[name_b], alternative="greater"), index=["stat", "pval"]), axis=1
    )
    ranksum_res["FDR"] = scipy.stats.false_discovery_control(ranksum_res["pval"])
    #record the means and their difference
    ranksum_res[name_a+" Mean"] = aligned[name_a].apply(np.mean)
    ranksum_res[name_b+" Mean"] = aligned[name_b].apply(np.mean)
    ranksum_res["Diff"] = ranksum_res[name_a+" Mean"] - ranksum_res[name_b+" Mean"]
    #call significant DEGs with effect size threshold
    ranksum_res["DEG"] = (ranksum_res["FDR"] < fdr_thresh) & (ranksum_res["Diff"] > diff_thresh)
    return ranksum_res, ranksum_res.query("DEG").index

orig, orig_sig = find_degs(
    exposed_skin_name, exposed_skin.groupby("Description").mean(), 
    protected_skin_name, protected_skin.groupby("Description").mean()
)
#misnormalize by increasing mito gene expr
more_mt = lambda x: x.multiply(np.where(x.index.str.startswith("MT-"), 5, 1), axis=0)
misnorm, misnorm_sig = find_degs(
    exposed_skin_name, exposed_skin.groupby("Description").mean(), 
    protected_skin_name, more_mt(protected_skin.groupby("Description").mean())
)
```

# LSA on GeneTEA embedding

```python
gene_cos = pd.DataFrame(cosine_similarity(tea.lsa_emb.loc[protein_coding]), index=protein_coding, columns=protein_coding)
```

```python
lsa_emb = tea.lsa_emb
real_mean, null_mean = {}, {}
rng = np.random.default_rng(27)
for i, real_group in hgnc_groups.items():
    real_mean[i] = tril_stack(gene_cos.loc[real_group, real_group]).mean()
    null_group = protein_coding.sample(len(real_group), random_state=rng.bit_generator)
    null_mean[i] = tril_stack(gene_cos.loc[null_group, null_group]).mean()
```

```python
group_labels = ["HGNC\nGene Groups", "Random\nGroups"]
mean_group_cos = pd.DataFrame([real_mean, null_mean], index=group_labels).T.stack().reset_index()

groups = mean_group_cos.groupby("level_1")[0].apply(list)
cos_pval = scipy.stats.ttest_ind(groups[group_labels[0]], groups[group_labels[1]], alternative="greater")
cos_pval = pd.Series({
    "statistic":cos_pval.statistic, "pvalue":cos_pval.pvalue, "df":cos_pval.df, 
    "n HGNC":len(groups[group_labels[0]]), "n Random":len(groups[group_labels[1]])
}).to_frame(tuple(group_labels)).T
cos_pval["str"] = "p="+cos_pval["pvalue"].apply(scientific_format)

plt.figure(figsize=(1.75,1.5), dpi=300)
distplot(
    data=mean_group_cos, x="level_1", y=0, hue="level_1", order=group_labels,
    hue_order=group_labels, palette=[PALETTE[7], PALETTE[6]], s=2, 
)
pairs = [(i[0], i[1]) for i in cos_pval.index]
annot_wrapper(
    plt.gca(), pairs, cos_pval, data=mean_group_cos, x="level_1", y=0, hue="level_1", 
    order=group_labels, hue_order=group_labels,
)
plt.legend().remove()
plt.xlabel(None)
plt.ylabel("Mean Cos. Sim.")
plt.grid(axis="y", alpha=0.2)

plt.savefig("../../figure-panels/Fig3c.jpg",bbox_inches='tight')
plt.show()
cos_pval
```
